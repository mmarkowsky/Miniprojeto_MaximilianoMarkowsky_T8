* Criação da seguinte função "diagnosticar_dataframe", a mesma vai gerar um relatório completo do diagnóstico e estrutura do DataFrame.A função vai receber como parâmetro um DF e tem como intuito analisar os seguintes pontos:

1.   Quantidade de filas e colunas utilizando a função shape()
2.   Quantidade de registros duplicados utilizando a função duplicated()
3.   Quantidade de registros nulos, null ,com espaços ou apenas aspas utilizando a função isnull() e expressões regulares
4.   Exibição transposta dos primeiros 5 registros do DF utilizando a função head()
5.   Exibição das quantidades de registros tipos de dados , valores nulos por coluna utilizando a função info()
6.   Exibição dos nomes das colunas utilizando columns.to_list()
7.   Contagem de valores negativos para colunas IDs
8.   Verificar valores unicos na coluna CL_GENERO

```python
def diagnosticar_dataframe(df):
    """
    Gera um relatório completo de diagnóstico e estrutura de um DataFrame.
    """
    print("="*60)
    print("📋 RELATÓRIO DE DIAGNÓSTICO DO DATAFRAME")
    print("="*60)

    # 1. Estrutura (Linhas e Colunas)
    linhas, colunas = df.shape
    print(f"\n 📊 Dimensões: {linhas} linhas e {colunas} colunas.")
    print("\n" + "-"*60 + "\n")

    # 2. Contagem de registros duplicados
    total_duplicados = df.duplicated().sum()
    porcentagem_duplicados = ((total_duplicados * 100) / len(df)).round(2)
    print(f"👥 Registros duplicados: {total_duplicados}.")
    print(f"👥 Porcentagem de registros duplicados: {porcentagem_duplicados}%.")
    print("\n" + "-"*60 + "\n")

    # 3. Análise detalhada por coluna (Nulos e Espaços vazios)
    nulos_reais = df.isnull().sum()
    porcentagem_nulos_reais = ((nulos_reais * 100) / len(df)).round(2)
    valores_invalidos = ['', ' ', '#N/D','NULL','N/D','undefined']
    colunas_valores_invalidos = df_varejo.isin(valores_invalidos).sum()
    porcentagem_valores_invalidos= ((colunas_valores_invalidos * 100) / len(df)).round(2)

    # Criar um DataFrame descritivo das colunas
    df_analise_cols = pd.DataFrame({
        'Valores Nulos (NaN)': nulos_reais,
        'Porcentagem Valores Nulos (NaN)':porcentagem_nulos_reais,
        'Contagem valores invalidos': colunas_valores_invalidos,
        'Porcentagem valores invalidos':porcentagem_valores_invalidos
    })

    print("🔍 Registros Nulos , Valores invalidos:")
    display(df_analise_cols)
    print("\n" + "-"*60 + "\n")

    # 4. Exibição das primeiras 5 filas do data frame
    print("🔄 Visualização transposta das primeiras 5 filas do dataframe:\n")
    df_transposto = df.head(5).reset_index().T
    display(df_transposto)
    print("\n" + "-"*60 + "\n")

    # 5. Estrutura do data frame com informações de Quantidade de registros por coluna ,Tipo de dado da coluna ,Valores nulos por coluna:
    print("🔬 Visualização do data frame com informações por coluna e tipo:\n")
    df_transposto = df.info()
    display(df_transposto)
    print("\n" + "-"*60 + "\n")

    # 6. Nomes das colunas
    print("🚀 Visualização dos nomes das colunas:\n")
    display(df_varejo.columns.to_list())
    print("\n" + "-"*60 + "\n")

    # 7. Contagem de valores negativos para colunas IDs
    print("🧮 Quantidade de valores negativos nas colunas dos IDs:\n")
    df_numerico = df_varejo.select_dtypes(include=['number'])
    mascara_negativos = df_numerico < 0
    display(mascara_negativos.sum())
    print("\n" + "-"*60 + "\n")

    # 8. Verificar valores unicos na coluna CL_GENERO
    print("📄 Visualização dos valores unicos da coluna CL_GENERO:\n")
    valores = df_varejo['CL_GENERO'].unique()
    print(f"Coluna CL_GENERO': {valores}")
    print("\n" + "="*60 + "\n")
```

* Padronizar os nomes para que não sejam considerados diferentes no momento da analise, foi feita uma função para limpar e normalizar strings (minúsculas, sem acentos/especiais),colunas afetadas 'PR_CAT'	e 'PR_NOME'.

```python
import unicodedata

def normalizar_texto(texto):
  # Garante que é string, remove espaços extras e deixa em minúscula
  texto_str = str(texto).strip().lower()

  # Remove acentos
  texto_sem_acentos = ''.join(
      c for c in unicodedata.normalize('NFD', texto_str)
      if unicodedata.category(c) != 'Mn'
  )

  # Mantém apenas letras, números e espaços simples
  texto_limpo = ''.join(c for c in texto_sem_acentos if c.isalnum() or c.isspace())
  return texto_limpo

df_varejo_limpo['PR_CAT'] = df_varejo_limpo['PR_CAT'].apply(normalizar_texto)
df_varejo_limpo['PR_NOME'] = df_varejo_limpo['PR_NOME'].apply(normalizar_texto)
df_varejo_limpo['CL_SEG'] = df_varejo_limpo['CL_SEG'].apply(normalizar_texto)
df_varejo_limpo['CL_GENERO'] = df_varejo_limpo['CL_GENERO'].apply(normalizar_texto)

display(df_varejo_limpo.head(5))
```
