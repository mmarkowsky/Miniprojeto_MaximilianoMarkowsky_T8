* Criação da seguinte função "diagnosticar_dataframe", a mesma vai gerar um relatório completo do diagnóstico e estrutura do DataFrame.A função vai receber como parâmetro um DF e tem como intuito analisar os seguintes pontos:

1.   Quantidade de filas e colunas utilizando a função shape()
2.   Quantidade de registros duplicados utilizando a função duplicated()
3.   Quantidade de registros nulos, null ,com espaços ou apenas aspas utilizando a função isnull() e expressões regulares
4.   Exibição transposta dos primeiros 5 registros do DF utilizando a função head()
5.   Exibição das quantidades de registros tipos de dados , valores nulos por coluna utilizando a função info()
6.   Exibição dos nomes das colunas utilizando columns.to_list()
7.   Contagem de valores negativos para colunas IDs
8.   Verificar valores unicos na coluna CL_GENERO

```python
def diagnosticar_dataframe(df):
    """
    Gera um relatório completo de diagnóstico e estrutura de um DataFrame.
    """
    print("="*60)
    print("📋 RELATÓRIO DE DIAGNÓSTICO DO DATAFRAME")
    print("="*60)

    # 1. Estrutura (Linhas e Colunas)
    linhas, colunas = df.shape
    print(f"\n 📊 Dimensões: {linhas} linhas e {colunas} colunas.")
    print("\n" + "-"*60 + "\n")

    # 2. Contagem de registros duplicados
    total_duplicados = df.duplicated().sum()
    porcentagem_duplicados = ((total_duplicados * 100) / len(df)).round(2)
    print(f"👥 Registros duplicados: {total_duplicados}.")
    print(f"👥 Porcentagem de registros duplicados: {porcentagem_duplicados}%.")
    print("\n" + "-"*60 + "\n")

    # 3. Análise detalhada por coluna (Nulos e Espaços vazios)
    nulos_reais = df.isnull().sum()
    porcentagem_nulos_reais = ((nulos_reais * 100) / len(df)).round(2)
    valores_invalidos = ['', ' ', '#N/D','NULL','N/D','undefined']
    colunas_valores_invalidos = df_varejo.isin(valores_invalidos).sum()
    porcentagem_valores_invalidos= ((colunas_valores_invalidos * 100) / len(df)).round(2)

    # Criar um DataFrame descritivo das colunas
    df_analise_cols = pd.DataFrame({
        'Valores Nulos (NaN)': nulos_reais,
        'Porcentagem Valores Nulos (NaN)':porcentagem_nulos_reais,
        'Contagem valores invalidos': colunas_valores_invalidos,
        'Porcentagem valores invalidos':porcentagem_valores_invalidos
    })

    print("🔍 Registros Nulos , Valores invalidos:")
    display(df_analise_cols)
    print("\n" + "-"*60 + "\n")

    # 4. Exibição das primeiras 5 filas do data frame
    print("🔄 Visualização transposta das primeiras 5 filas do dataframe:\n")
    df_transposto = df.head(5).reset_index().T
    display(df_transposto)
    print("\n" + "-"*60 + "\n")

    # 5. Estrutura do data frame com informações de Quantidade de registros por coluna ,Tipo de dado da coluna ,Valores nulos por coluna:
    print("🔬 Visualização do data frame com informações por coluna e tipo:\n")
    df_transposto = df.info()
    display(df_transposto)
    print("\n" + "-"*60 + "\n")

    # 6. Nomes das colunas
    print("🚀 Visualização dos nomes das colunas:\n")
    display(df_varejo.columns.to_list())
    print("\n" + "-"*60 + "\n")

    # 7. Contagem de valores negativos para colunas IDs
    print("🧮 Quantidade de valores negativos nas colunas dos IDs:\n")
    df_numerico = df_varejo.select_dtypes(include=['number'])
    mascara_negativos = df_numerico < 0
    display(mascara_negativos.sum())
    print("\n" + "-"*60 + "\n")

    # 8. Verificar valores unicos na coluna CL_GENERO
    print("📄 Visualização dos valores unicos da coluna CL_GENERO:\n")
    valores = df_varejo['CL_GENERO'].unique()
    print(f"Coluna CL_GENERO': {valores}")
    print("\n" + "="*60 + "\n")
```

* Padronizar os nomes para que não sejam considerados diferentes no momento da analise, foi feita uma função para limpar e normalizar strings (minúsculas, sem acentos/especiais),colunas afetadas 'PR_CAT'	e 'PR_NOME'.

```python
import unicodedata

def normalizar_texto(texto):
  # Garante que é string, remove espaços extras e deixa em minúscula
  texto_str = str(texto).strip().lower()

  # Remove acentos
  texto_sem_acentos = ''.join(
      c for c in unicodedata.normalize('NFD', texto_str)
      if unicodedata.category(c) != 'Mn'
  )

  # Mantém apenas letras, números e espaços simples
  texto_limpo = ''.join(c for c in texto_sem_acentos if c.isalnum() or c.isspace())
  return texto_limpo

df_varejo_limpo['PR_CAT'] = df_varejo_limpo['PR_CAT'].apply(normalizar_texto)
df_varejo_limpo['PR_NOME'] = df_varejo_limpo['PR_NOME'].apply(normalizar_texto)
df_varejo_limpo['CL_SEG'] = df_varejo_limpo['CL_SEG'].apply(normalizar_texto)
df_varejo_limpo['CL_GENERO'] = df_varejo_limpo['CL_GENERO'].apply(normalizar_texto)

display(df_varejo_limpo.head(5))
```

# Sprint 3 (Funções e organização)
* Criar e chamar uma função de tratamento com parâmetros e retorno.
* Criar e chamar uma função de resumo com parâmetros e retorno.
* Manter uma sequência clara entre leitura, limpeza e análise.

* Foram criadas 3 funções:

1. normalizar_texto(): Vai receber um texto e o mesmo vai ser padronizado em minuscula , sem espaços extras e sem caracteres especias retornando um texto limpo
2. tratamento_dados(): Vai excluir de registros nulos e registros duplicados, padronizar dados das colunas de tipo string (utilizando a primeira função normalizar_texto()) e formatar campo data
3. tratamento_resumo(): Vai mostrar um relatorio de forma clara com as modificações e diferenças entre o DF original e DF limpo

```python
import numpy as np
import unicodedata

def normalizar_texto(texto):
  # Garante que é string, remove espaços extras e deixa em minúscula
  texto_str = str(texto).strip().lower()

  # Remove acentos
  texto_sem_acentos = ''.join(
      c for c in unicodedata.normalize('NFD', texto_str)
      if unicodedata.category(c) != 'Mn'
  )

  # Mantém apenas letras, números e espaços simples
  texto_limpo = ''.join(c for c in texto_sem_acentos if c.isalnum() or c.isspace())
  return texto_limpo


def tratamento_dados(df):
  # Deletando colunas com mais do 80% de valores nulos
  df = df.drop(columns=['Unnamed: 10','Unnamed: 11','Unnamed: 12','Unnamed: 13'])

  # Deletando filas das colunas 'PR_CAT','PR_NOME' com valores nulos
  df = df.replace(['#N/D'], np.nan)
  df.dropna(subset=['PR_CAT', 'PR_NOME'],inplace=True)

  # Deletando registros duplicados mantendo a primeira aparição com o parâmetro (keep='first')
  df = df.drop_duplicates(keep='first')

  #Reindexando os indices depois da exclusão dos registros
  df = df.reset_index(drop=True)

  #Modificar o formato da coluna DATA para o tipo DATETIME
  df['DATA'] = pd.to_datetime(
      df['DATA'],
      format='%d/%m/%Y'
  )

  #Filtrando apenas colunas de tipo String e Object
  df_colunas_string = df.select_dtypes(include=['object', 'string'])
  #Percorrer o dataframe naquelas colunas(tipo Object e String) para normalizar o texto")
  for coluna in df_colunas_string.columns:
    df[coluna] = df[coluna].apply(normalizar_texto)

  return df


def tratamento_resumo(df_original, df_copia):
  # 1. Cálculos de métricas
  tamanho_original = len(df_original)
  tamanho_limpo = len(df_copia)
  deletados = tamanho_original - tamanho_limpo

  # Evita divisão por zero caso o DataFrame original esteja vazio
  porcentagem_deletados = (deletados * 100) / tamanho_original

  # 2. Exibição do Relatório Estruturado
  print("=" * 60)
  print("📋 RELATÓRIO DE TRATAMENTO E LIMPEZA DE DADOS".center(60))
  print("=" * 60)

  print(f"🔹 Tamanho do DataFrame original        : {tamanho_original:,} linhas")
  print(f"🔹 Tamanho do DataFrame pós-limpeza     : {tamanho_limpo:,} linhas")
  print(f"❌ Quantidade de registros deletados    : {deletados:,} ({porcentagem_deletados:.2f}%)")

  print("-" * 60)
  print("📌 TIPOS DE COLUNAS ALTERADAS:")
  print(f"   • Tipo atual da coluna 'DATA'        : {df_copia['DATA'].dtypes}")

  print("-" * 60)
  print("📌 COLUNAS NORMALIZADAS:")
  df_colunas_string = df_copia.select_dtypes(include=['object', 'string'])
  for coluna in df_colunas_string.columns:
    print(f"   • Coluna        : {coluna}")

  print("-" * 60)
  print("👀 VISUALIZAÇÃO DOS PRIMEIROS REGISTROS (PÓS-LIMPEZA):\n")
  # 3. Renderização dos dados (Usa o df_copia recebido no argumento)
  display(df_copia.head(5))
  print("=" * 60)
```

* Aplicar o tratamento na copia do arquivo original utilizando a função tratamento_dados()

```python
df_varejo_limpo = tratamento_dados(df_varejo_limpo)
```

* Visualizar o resumo do tratamento utilizando a função tratamento_resumo()

* Criação de um filtro por quantidade de compras por cada clientes. Logo vai ser implementada uma regra para definir o tipo de cliente em uma nova coluna:

```python
#Contagem de compras por clientes, salvando em uma variavel compras_x_cliente
compras_x_cliente = df_varejo_limpo['CL_ID'].value_counts()

#Aplicando uma função lambda para definir o tipo de cliente para cada "CL_ID" do meu DF mapeando com a minha variavel compras_x_cliente
df_varejo_limpo['CL_TIPO'] = df_varejo_limpo['CL_ID'].map(compras_x_cliente).apply(lambda x: 'vip' if x >= 1500 else 'recurrente' if 500 < x & x < 1500 else 'eventual')

display(df_varejo_limpo['CL_TIPO'].value_counts())
```

* Calculo da quantidade COMPRAS em relação ao campo DATA

```python
#Criação uma nova coluna apenas com o ano
df_varejo_limpo['ANO'] = df_varejo_limpo['DATA'].dt.year

#Agrupa por Cliente (CO_ID) e Ano
compras_x_ano = df_varejo_limpo.groupby('ANO')['CO_ID'].agg(
    Quantidade_Por_Ano='count',          # Conta quantas compras foram feitas por ano
).reset_index()

display(compras_x_ano)
```

* Criação de um groupby para responder a seguinte questão:

Questão: Saber quais são os 5 produtos mais vendidos agrupando por GENERO e ESTADO CIVIL

```python
#Contagem das vendas por NOME PRODUTO agrupando por ESTADO CIVIL E GENERO
top_produtos_ec_genero = (df_varejo_limpo.groupby(['CL_EC', 'CL_GENERO'])['PR_NOME']
                           .value_counts()
                           .reset_index(name='QUANTIDADE'))

#Filtragem dos primeiros 5 produtos mais vendidos para cada Estado Civil e Genero
top_5_completo = top_produtos_ec_genero.groupby(['CL_EC', 'CL_GENERO']).head(5)

#Mostrar resultado
display(top_5_completo.style.background_gradient(cmap='Blues',axis=None))
```
